### Математические операторы
"+" - сложение

"-" - вычитание

"/" - деление

"//" - целочисленное деление

"%" - остаток от деления

"*" - умножение

"**" - возведение в степень

````python
a = 7
b = 2
print("a + b = ", a + b)
print("a - b = ", a - b)
print("a / b = ", a / b)
print("a // b = ", a // b)
print("a % b = ", a % b) # В данном случае ближайшее число к 7, которое делится на 2 без остатка, это 6. Поэтому остаток от деления равен 7 - 6 = 1
print("a * b = ", a * b)
print("a ** b = ", a ** b)
````

#### Арифметические операции с присвоением
Ряд специальных операций позволяют использовать присвоить результат операции первому операнду:

"+=" - Присвоение результата сложения

"-=" - Присвоение результата вычитания

"*=" - Присвоение результата умножения

"/=" - Присвоение результата от деления

"//=" - Присвоение результата целочисленного деления

"**=" - Присвоение степени числа

"%=" - Присвоение остатка от деления

### Логические операторы

"==" - Возвращает True, если оба операнда равны. Иначе возвращает False.

"!=" - Возвращает True, если оба операнда НЕ равны. Иначе возвращает False.

"> (больше чем)" - Возвращает True, если первый операнд больше второго.

"< (меньше чем)" - Возвращает True, если первый операнд меньше второго.

">= (больше или равно)" - Возвращает True, если первый операнд больше или равен второму.

"<= (меньше или равно)" - Возвращает True, если первый операнд меньше или равен второму.

"and" - логическое умножение

"or" - логическое сложение

"not" - логическое отрицание

````python
print("a > b =",a > b)
print("a < b = ",a > b)
print("a == b = ",a == b)
print("a != b = ",a != b)
print("a >= b = ",a >= b)
print("a < b and a > b = ", a < b and a > b)
print("a < b or a > b = ", a < b or a > b)
print("not a < b = ", not a < b)
````

### Лямбда функции

Лямбда-выражения в языке Python представляют небольшие анонимные функции,
которые определяются с помощью оператора lambda. Формальное определение
лямбда-выражения: `lambda [параметры] : инструкция`

`message = lambda: print("hello")`
 
Вызов `message()` выведет сообщение hello

Здесь лямбда-выражение присваивается переменной message. Это лямбда-выражение не
имеет параметров, ничего не возвращает и просто выводит строку "hello" на
консоль. И через переменную message мы можем вызвать это лямбда-выражение как
обычную функцию. Фактически оно аналогично следующей функции:

`def message(): `
    `print("hello")`

````python
square = lambda n: n * n

print(square(4))
print(square(5))
````

Хотя лямбда-выражения позволяют немного сократить определения функций, тем не
менее они ограничены тем, что они могут выполнять только одну инструкцию. Однако
они могут быть довольно удобны в тех случаях, когда необходимо использовать
функцию для передачи в качестве параметра или возвращения в другой функции. 

````python
def do_operation(a, b, operation):
    result = operation(a, b)
    print(f"result = {result}")

do_operation(5, 4, lambda a, b: a + b)
do_operation(5, 4, lambda a, b: a * b)
````

## Работа с файлами

Python позволяет оперировать разными типами документов: текстовыми в любом
формате, графическими, медиа, табличными и json. Для этого имеются
соответствующие модули и пакеты (так встроенные, так и те, которые необходимо
дополнительно установить), а также ряд функций.

При работе с файлами в Python используется ряд функций и методов:
- функция `open()` - открывает файл для чтения, записи, добавления нового
  содержимого. Может принимать дополнительные параметры: для задания режима
  открытия, указания кодировки, вывода ошибок и др. Возвращает дескриптор файла,
  который обязательно нужно закрыть, иначе файл останется в памяти. Дескриптор в
  данном случае представляет собой путь к документу в виде строки;
- `close()` - закрывает файловый объект;
- инструкция `with` (позволяет автоматически закрывать файловый объект после
  работы с ним);
- метод `read()` - для чтения содержимого документа;
- метод `readlines()` - преобразует все строки файла в список;
- метод `readline()` - построчно выводит данные файла (удобно при работе с
  массивными документами);
- метод `ewrit()` - записывает новую информацию в файл;
- функция `rename()` из модуля os - переименовывает документ и др.


`f = open('text.txt', 'r')` - открытие текстового файла для чтения

`img = Image.open('pic.png')` - открываем PNG

##### Скалярное произведение и угол между векторами

Скалярное произведение в пространстве $\mathbb{R}^{n}$ для двух векторов $x =
(x_{1}, \dots, x_{n})$ и $y = (y_{1}, \dots, y_{n})$ определяется как:

$$ \langle x, y \rangle = \sum_{i=1}^n x_{i} y_{i}. $$

Скалярное произведение двух векторов можно вычислять с помощью функции
`numpy.dot(a, b, ...)` или метода `vec1.dot(vec2)`, где `vec1` и `vec2` —
исходные векторы. Также эти функции подходят для матричного умножения.

Длиной вектора $x = (x_{1}, \dots, x_{n}) \in \mathbb{R}^{n}$ называется
квадратный корень из скалярного произведения, то есть длина равна евклидовой
норме вектора:

$$ \left| x \right| = \sqrt{\langle x, x \rangle} = \sqrt{\sum_{i=1}^n x_{i}^2} = \left\Vert x \right\Vert_{2}. $$

Теперь, когда мы знаем расстояние между двумя ненулевыми векторами и их длины,
мы можем вычислить угол между ними через скалярное произведение:

$$ \langle x, y \rangle = \left| x \right| | y | \cos(\alpha) \implies \cos(\alpha) = \frac{\langle x, y \rangle}{\left| x \right| | y |}, $$
где $\alpha \in [0, \pi]$ — угол между векторами $x$ и $y$.

````python
a = np.array([0, 5, -1])
b = np.array([-4, 9, 3])
print(f'Вектор a:{a}\nВектор b: {b}\nСкалярное произведение a и b np.dot(a, b) = {np.dot(a, b)}\nСкалярное произведение a и b a.dot(b) = {a.dot(b)}')

cos_angle = np.dot(a, b) / norm(a) / norm(b)
print(f'Косинус угла между a и b cos_angle = {cos_angle}\nСам угол np.arccos(cos_angle) = {np.arccos(cos_angle)}')
````

##### Расстояния между векторами

Для двух векторов $x = (x_{1}, \dots, x_{n}) \in \mathbb{R}^{n}$ и $y = (y_{1},
\dots, y_{n}) \in \mathbb{R}^{n}$ $\ell_{1}$ и $\ell_{2}$ расстояния вычисляются
по следующим формулам соответственно:

$$ \rho_{1}\left( x, y \right) = \left\Vert x - y \right\Vert_{1} = \sum_{i=1}^n \left| x_{i} - y_{i} \right| $$
$$ \rho_{2}\left( x, y \right) = \left\Vert x - y \right\Vert_{2} = \sqrt{\sum_{i=1}^n \left( x_{i} - y_{i} \right)^2}. $$


````python
from numpy.linalg import norm


a = np.array([1, 2, -3])
b = np.array([-4, 3, 8])
print(f'Вектор a:{a},\nВектор b:{b},\nРасстояние между векторами a и b norm(a - b, ord=1) = {norm(a - b, ord=1)}')
````

Функция `reshape()` изменяет форму массива без изменения его данных.

**Важно:** Новая форма должна содержать то же количество элементов, что и исходная.

```python
numpy.reshape(np_array, new_shape, order, order='C')
```

Параметры:
- np_array - исходный массив
- new_shape - новая форма (целое число или кортеж)
- order - порядок элементов, возможные значения: 'C', 'F', 'A':
  - `order='C'`. Упорядочивание индексов в стиле языка C. Индекс последней оси изменяется быстрее, а индекс первой — медленнее. Индексация идет слева направо, потом перемещается на вторую строку и снова идет слева направо, дальше перемещается на третью.
  - `order='F'`. Упорядочивания индексов в стиле языка Фортран. Индекс первой оси изменяется быстрее, а индекс последней — медленнее. Индексация идет в первом столбце сверху вниз, потом переходит на второй столбец и снова идет сверху вниз, потому (аналогично) - на третий и четвертый столбцы.
  - `order='A'`. Варианты «C» и «F» не учитывают макет памяти основного массива. Они относятся лишь к порядку индексации. Порядок «A» означает чтение и запись элементов в стиле Фортран, если исходный массив в памяти тоже в стиле Фортран. В противном случае применяется C-подобный стиль.

Отличие от numpy.resize():

- `resize()` меняет размер массива, добавляя/удаляя элементы при необходимости
- `reshape()` требует точного соответствия размеров

````python
# Преобразование одномерного массива в двумерный
np_array = np.arange(6)
new_arr = np.reshape(np_array, (2, 3))
print(f'Исходный массив =\n {np_array}, \n Новый массив = \n {new_arr}')

# Преобразование одномерного массива в трехмерный
np_array = np.arange(24)
new_arr = np.reshape(np_array, (2, 3, 4))
print(f'Исходный массив =\n {np_array}, \n Новый массив = \n {new_arr}')

# Пример с двумя массивами
v = np.array([1,2,3])
w = np.array([4,5])
new_arr = np.reshape(v, (3, 1)) * w
print(f'Исходный массив =\n {v}, \n Новый массив = \n {new_arr}')
````

Расчет дифференциального уравнения второго порядка для угла $\theta$ маятника,
на который действует сила тяжести с трением: $\theta''(t) + b*\theta'(t) +
c*sin(\theta(t)) = 0$, $b$ и $c$ — положительные константы

Данное уравнение можно представить как систему уравнений

$\theta'(t) = \omega(t)$

$\omega'(t) = -b*\omega(t) - c*\sin(\theta(t))$

Пусть $y$ будет вектором $\{\theta, \omega\}$, тогда решение:

````python
from scipy.integrate import odeint


def pend(y, t, b, c):
    theta, omega = y
    dydt = [omega, -b*omega - c*np.sin(theta)]
    return dydt


b = 0.25
c = 5.0
y0 = [np.pi - 0.1, 0.0]
t = np.linspace(0, 10, 101)

sol = odeint(pend, y0, t, args=(b, c))

plt.plot(t, sol[:, 0], 'b', label='theta(t)')
plt.plot(t, sol[:, 1], 'g', label='omega(t)')
plt.legend(loc='best')
plt.xlabel('t')
plt.grid()
plt.show()
````

### copy

Файлы, сокеты, потоки - копируются только ссылки

````python
# Поверхностное копирование (copy.copy()). Создает новый объект, но вставляет ссылки на вложенные объекты оригинала.

original = [1, [2, 3], 4]
shallow = copy.copy(original)
print(f'Пример для списка:\nБыл список {original}, сделали копию {shallow}')
original[1].append(5)
print(f'Внесли изменения, изменилась копия {shallow[1]}')

d = {'a': [1, 2], 'b': 3}
d_copy = copy.copy(d)

print(f'Пример для словаря:\nБыл словарь {d}, сделали копию {d_copy}')

d['a'].append(4)
print(d_copy['a'])  # [1, 2, 4]

print('Внесли изменения, изменилась копия')
print(d_copy['a'])  # [1, 2, 4]
````

````python
# Глубокое копирование (copy.deepcopy()). Рекурсивно создает полностью независимую копию объекта.

original = [1, [2, 3], {'a': 4}]
deep = copy.deepcopy(original)
print(f'Вложенные списки:\nБыло {original}, копия {deep}')

original[1].append(5)
print(deep[1])  # [2, 3] - копия не изменилась
print(f'Внесли изменения, изменилась копия {deep[1]}')
````

Официальная документация:

### Что такое ООП

Объектно-ориентированное программирование (ООП) — парадигма, где программа строится из объектов, которые взаимодействуют друг с другом.

#### Основные понятия

- **Класс** - это шаблон или чертеж, который определяет свойства и поведение
  объекта. Классы в Python определяются с помощью ключевого слова `class`.
- **Объект** - это экземпляр класса. Объекты имеют свои собственные свойства и
  поведение, которые наследуются от класса.
- **Атрибуты** - это данные, которые хранятся в объекте (переменные). Атрибуты
  могут быть доступны через точечную нотацию.
- **Методы** - это функции, которые принадлежат объекту (функции). Методы могут
  быть использованы для изменения состояния объекта или для выполнения
  определенных действий.
- **Конструктор** - это специальный метод, который вызывается при создании
  объекта. Конструктор используется для инициализации атрибутов объекта. В
  Python конструктор определяется с помощью метода `__init__`.
- **Методы для переопределения поведения объекта** методы вызываются
  автоматически при выполнении определенных операций, таких как сравнение
  объектов или преобразование объекта в строку. Примеры методов с двойным
  подчеркиванием включают `__str__`, `__repr__`, `__eq__` и `__lt__`.
- **Self** - это ссылка на текущий объект, которая передается в методы класса.
  `self` используется для доступа к атрибутам и методам объекта изнутри метода.
- **Декораторы** - это специальные функции, которые могут изменять или расширять
  поведение других функций. Декораторы определяются с помощью ключевого слова
  `@` и могут быть использованы для добавления функциональности к методам
  класса.
- **Динамическое создание атрибутов** - это процесс создания атрибутов объекта
  во время выполнения программы, а не во время определения класса. В Python это
  можно сделать с помощью нескольких методов:
  - Оператор `setattr()`
```python
obj = MyClass()
setattr(obj,'my_attr', 'Hello, world!')
print(obj.my_attr)  # Вывод: Hello, world!
```
  - Точечно
```python
obj = MyClass()
obj.my_attr = 'Hello, world!'
print(obj.my_attr)  # Вывод: Hello, world!
```
  - Метод `__dict__`
```python
obj = MyClass()
obj.dict['my_attr'] = 'Hello, world!'
print(obj.my_attr)  # Вывод: Hello, world!
```

##### Пример

```python
# Декоратор
def my_decorator(func):
    def wrapper(self):
        print("Декоратор вызван")
        func(self)
    return wrapper

# Класс
class Transport:
    # Конструктор
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    # Метод
    @my_decorator
    def move(self):
        print("Транспорт движется")

    def __str__(self):
        return f"{self.brand} {self.model}"

    def __repr__(self):
        return f"Transport({self.brand}, {self.model})"

# Создание объекта
my_car = Transport("Toyota", "Corolla")

# Динамическое создание атрибута
setattr(my_car, 'year', 2015)

# Доступ к атрибутам и методам
print(my_car.brand)  # Вывод: Toyota
print(my_car.model)  # Вывод: Corolla
print(my_car.year)   # Вывод: 2015
my_car.move()        # Вывод: Декоратор вызван, Транспорт движется
print(my_car)        # Вывод: Toyota Corolla
print(repr(my_car))  # Вывод: Transport(Toyota, Corolla)


class Car(Transport):
    def __init__(self, brand, model, year):
        super().__init__(brand, model)
        self.year = year

    def print_info(self):
        print(f"Марка: {self.brand}, Модель: {self.model}, Год: {self.year}")

my_car = Car("Toyota", "Corolla", 2015)
my_car.print_info()  # Вывод: Марка: Toyota, Модель: Corolla, Год: 2015
```

#### Основные принципы ООП

1. Инкапсуляция - это механизм, который скрывает внутреннюю реализацию объекта от
   внешнего мира. В Python инкапсуляция может быть достигнута с помощью
   приватных атрибутов и методов.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0  # Приватный атрибут

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance
```

2. Наследование - это механизм, который позволяет создавать новые классы на
   основе существующих классов. Новый класс наследует все атрибуты и методы
   родительского класса.

```python
class Terrier(Dog):  # Наследование от Dog
    def hunt(self):
        print(f"{self.name} is hunting rats!")
```

3. Полиморфизм - это способность объекта принимать разные формы. В Python
   полиморфизм может быть достигнут с помощью методов, которые имеют одинаковые
   имена, но разные реализации.

```python
class Cat:
    def speak(self):
        print("Meow!")

class Duck:
    def speak(self):
        print("Quack!")

for animal in [Dog("Buddy", 2), Cat(), Duck()]:
    animal.speak()
```

### Почему ООП может быть избыточен для научных скриптов?

Простота скриптов для решения базовых задач. Научные вычисления часто
представляют собой линейные последовательности операций. В таком случае проще
написать скрипт, чем применять ООП. В таких скриптах акцент делается на данные,
а не объекты. Пример простого научного скрипта:

```python
import numpy as np
data = np.loadtxt("data.csv")
result = np.mean(data) * 100
print(f"Result: {result:.2f}%")
```

В научных задачах чаще работают с:

* Массивами данных (NumPy)
* Матрицами (SciPy)

В научных вычислениях используется процедурный стиль. Многие научные библиотеки
уже инкапсулируют сложность:

```python
# Использование функций вместо классов
from scipy import stats
t_stat, p_value = stats.ttest_ind(group_a, group_b)
```

Используйте функциональный подход:

```python
def calculate_entropy(signal):
    histogram = np.histogram(signal, bins=50)[0]
    probabilities = histogram / len(signal)
    return -np.sum(probabilities * np.log2(probabilities))
```
Применяйте векторизацию вместо объектов:

```python
# Вместо класса для векторов:
vectors = np.array([[1, 2], [3, 4], [5, 6]])
magnitudes = np.linalg.norm(vectors, axis=1)
```

### Когда ООП всё же полезен в науке?

Создание классов добавляет:

- Дополнительный уровень абстракции
- Необходимость проектирования иерархий
- Потенциальные проблемы наследования

Когда в научных вычислениях стоит применять ООП:

- Моделирование сложных систем (например, агентное моделирование)
- Создание библиотек и инструментов для повторного использования
- Работа с иерархическими структурами данных
- Долгосрочные проекты с развивающейся архитектурой
- Рекомендации для научных скриптов
